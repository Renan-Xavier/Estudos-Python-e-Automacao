```python
# ===== LISTAS =====
# Mutáveis, ordenadas, permitem duplicatas
lista = [1, 2, 3, 4, 5]
mista = [1, "texto", 3.14, True]

print("LISTAS:")
print(f"Lista: {lista}")
print(f"Primeiro: {lista[0]}, Último: {lista[-1]}")
print(f"Slice [1:4]: {lista[1:4]}")

# Métodos
lista.append(6)          # Adiciona no final
lista.insert(0, 0)       # Insere na posição
lista.extend([7, 8])     # Adiciona múltiplos
lista.remove(3)          # Remove primeira ocorrência
ultimo = lista.pop()     # Remove e retorna último

print(f"Modificada: {lista}")
print(f"Tamanho: {len(lista)}")
print(f"Soma: {sum(lista)}")
print(f"Ordenada: {sorted(lista)}")

# ===== TUPLAS =====
# Imutáveis, ordenadas, permitem duplicatas
print("\nTUPLAS:")
tupla = (1, 2, 3, 4, 5)
coordenadas = (10, 20)

# Desempacotamento
x, y = coordenadas
print(f"x={x}, y={y}")

primeiro, *meio, ultimo = tupla
print(f"Primeiro: {primeiro}, Meio: {meio}, Último: {ultimo}")

# ===== DICIONÁRIOS =====
# Mutáveis, pares chave-valor, não ordenados (Python 3.7+ mantém ordem de inserção)
print("\nDICIONÁRIOS:")
pessoa = {
    'nome': 'João',
    'idade': 30,
    'cidade': 'São Paulo'
}

print(f"Pessoa: {pessoa}")
print(f"Nome: {pessoa['nome']}")
print(f"Email: {pessoa.get('email', 'Não informado')}")

pessoa['telefone'] = '123456789'  # Adiciona
pessoa['idade'] = 31               # Modifica

print(f"\nChaves: {list(pessoa.keys())}")
print(f"Valores: {list(pessoa.values())}")

print("\nIterando:")
for chave, valor in pessoa.items():
    print(f"{chave}: {valor}")

# ===== CONJUNTOS (SETS) =====
# Mutáveis, não ordenados, SEM duplicatas
print("\nCONJUNTOS:")
conjunto = {1, 2, 3, 4, 5}
com_duplicatas = {1, 2, 2, 3, 3, 3}
print(f"Conjunto: {conjunto}")
print(f"Duplicatas removidas: {com_duplicatas}")

# Operações
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}
print(f"\nA: {A}")
print(f"B: {B}")
print(f"União: {A | B}")
print(f"Interseção: {A & B}")
print(f"Diferença: {A - B}")
print(f"Diferença simétrica: {A ^ B}")
```

```python
# Função básica
def saudacao(nome):
    return f"Olá, {nome}!"

print(saudacao("Maria"))

# Múltiplos retornos
def operacoes(a, b):
    return a + b, a - b, a * b, a / b

soma, sub, mult, div = operacoes(10, 2)
print(f"\nOperações: {soma}, {sub}, {mult}, {div}")

# Parâmetros padrão
def potencia(base, exp=2):
    return base ** exp

print(f"\n5²: {potencia(5)}")
print(f"5³: {potencia(5, 3)}")

# *args e **kwargs
def soma_todos(*args):
    return sum(args)

print(f"\nSoma: {soma_todos(1, 2, 3, 4, 5)}")

def info(**kwargs):
    for k, v in kwargs.items():
        print(f"{k}: {v}")

print("\nInformações:")
info(nome="João", idade=30, cidade="SP")

# LAMBDA (funções anônimas)
quadrado = lambda x: x ** 2
print(f"\nQuadrado de 5: {quadrado(5)}")

# map, filter, sorted com lambda
nums = [1, 2, 3, 4, 5]
quadrados = list(map(lambda x: x**2, nums))
pares = list(filter(lambda x: x % 2 == 0, nums))

print(f"Quadrados: {quadrados}")
print(f"Pares: {pares}")

palavras = ['python', 'java', 'c', 'javascript']
ordenadas = sorted(palavras, key=lambda x: len(x))
print(f"Ordenadas por tamanho: {ordenadas}")

# DOCSTRINGS
def calcular_area(largura, altura):
    """
    Calcula área do retângulo.
    
    Args:
        largura: Largura do retângulo
        altura: Altura do retângulo
    
    Returns:
        Área calculada
    """
    return largura * altura

print(f"\nÁrea: {calcular_area(5, 3)}")
```

## 6. Programação Orientada a Objetos

```python
# Classe básica
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade
    
    def apresentar(self):
        return f"Olá, sou {self.nome}, {self.idade} anos"
    
    def aniversario(self):
        self.idade += 1

p1 = Pessoa("João", 30)
print(p1.apresentar())
p1.aniversario()
print(p1.apresentar())

# HERANÇA
class Animal:
    def __init__(self, nome):
        self.nome = nome
    
    def fazer_som(self):
        return "Som genérico"

class Cachorro(Animal):
    def fazer_som(self):
        return "Au au!"
    
    def buscar(self):
        return f"{self.nome} buscou a bola!"

class Gato(Animal):
    def fazer_som(self):
        return "Miau!"

dog = Cachorro("Rex")
cat = Gato("Mimi")

print(f"\n{dog.nome}: {dog.fazer_som()}")
print(f"{cat.nome}: {cat.fazer_som()}")
print(dog.buscar())

# ENCAPSULAMENTO
class ContaBancaria:
    def __init__(self, titular, saldo=0):
        self.titular = titular
        self.__saldo = saldo  # Atributo privado
    
    @property
    def saldo(self):
        return self.__saldo
    
    def depositar(self, valor):
        if valor > 0:
            self.__saldo += valor
            return f"Depositado: R${valor}"
        return "Valor inválido"
    
    def sacar(self, valor):
        if 0 < valor <= self.__saldo:
            self.__saldo -= valor
            return f"Sacado: R${valor}"
        return "Saldo insuficiente"
    
    def __str__(self):
        return f"Conta de {self.titular}: R${self.__saldo}"

conta = ContaBancaria("Maria", 1000)
print(f"\n{conta}")
print(conta.depositar(500))
print(conta.sacar(300))
print(f"Saldo: R${conta.saldo}")

# MÉTODOS ESPECIAIS (Magic Methods)
class Ponto:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __str__(self):
        return f"Ponto({self.x}, {self.y})"
    
    def __add__(self, outro):
        return Ponto(self.x + outro.x, self.y + outro.y)
    
    def __eq__(self, outro):
        return self.x == outro.x and self.y == outro.y

p1 = Ponto(1, 2)
p2 = Ponto(3, 4)
p3 = p1 + p2

print(f"\n{p1} + {p2} = {p3}")
print(f"p1 == p2: {p1 == p2}")
```

## 7. Arquivos e Exceções

```python
# MANIPULAÇÃO DE ARQUIVOS

# Escrevendo
with open('teste.txt', 'w', encoding='utf-8') as f:
    f.write("Linha 1\n")
    f.write("Linha 2\n")
    f.write("Linha 3\n")

print("Arquivo criado!")

# Lendo
with open('teste.txt', 'r', encoding='utf-8') as f:
    conteudo = f.read()
    print(f"\nConteúdo:\n{conteudo}")

# Lendo linha por linha
with open('teste.txt', 'r', encoding='utf-8') as f:
    for i, linha in enumerate(f, 1):
        print(f"Linha {i}: {linha.strip()}")

# Append (adicionar)
with open('teste.txt', 'a', encoding='utf-8') as f:
    f.write("Linha 4 (adicionada)\n")

# JSON
import json

dados = {
    'nome': 'João',
    'idade': 30,
    'hobbies': ['programar', 'ler']
}

# Salvar JSON
with open('dados.json', 'w') as f:
    json.dump(dados, f, indent=2)

# Ler JSON
with open('dados.json', 'r') as f:
    dados_lidos = json.load(f)
    print(f"\nJSON lido: {dados_lidos}")

# TRATAMENTO DE EXCEÇÕES

# Try-except básico
try:
    resultado = 10 / 0
except ZeroDivisionError:
    print("\nErro: Divisão por zero!")

# Múltiplas exceções
try:
    lista = [1, 2, 3]
    print(lista[10])
except (IndexError, KeyError) as e:
    print(f"Erro de acesso: {e}")

# Try-except-else-finally
try:
    arquivo = open('teste.txt', 'r')
except FileNotFoundError:
    print("Arquivo não encontrado")
else:
    print("Arquivo aberto com sucesso")
    arquivo.close()
finally:
    print("Finally sempre executa")

# Exceção personalizada
class ValorNegativoError(Exception):
    pass

def raiz_quadrada(n):
    if n < 0:
        raise ValorNegativoError("Não pode ser negativo")
    return n ** 0.5

try:
    print(f"\nRaiz de -4: {raiz_quadrada(-4)}")
except ValorNegativoError as e:
    print(f"Erro: {e}")
```

## 8. Tópicos Avançados

```python
# LIST COMPREHENSION
quadrados = [x**2 for x in range(10)]
print(f"Quadrados: {quadrados}")

pares = [x for x in range(20) if x % 2 == 0]
print(f"Pares: {pares}")

matriz = [[i*j for j in range(1, 4)] for i in range(1, 4)]
print(f"Matriz: {matriz}")

# DICT COMPREHENSION
quadrados_dict = {x: x**2 for x in range(5)}
print(f"\nDict quadrados: {quadrados_dict}")

# SET COMPREHENSION
quadrados_set = {x**2 for x in range(-5, 6)}
print(f"Set quadrados: {quadrados_set}")

# GENERATORS (economizam memória)
def contador(max):
    n = 0
    while n < max:
        yield n
        n += 1

print("\nGenerator:")
for num in contador(5):
    print(num, end=" ")

# Generator expression
gen = (x**2 for x in range(5))
print(f"\n\nGenerator expression: {list(gen)}")

# DECORATORS
def meu_decorador(func):
    def wrapper(*args, **kwargs):
        print("Antes da função")
        resultado = func(*args, **kwargs)
        print("Depois da função")
        return resultado
    return wrapper

@meu_decorador
def saudacao(nome):
    print(f"Olá, {nome}!")

print("\nDecorator:")
saudacao("Maria")

# CONTEXT MANAGERS
class MeuContexto:
    def __enter__(self):
        print("\nEntrando no contexto")
        return self
    
    def __exit__(self, exc_type, exc_val, exc_tb):
        print("Saindo do contexto")

with MeuContexto():
    print("Dentro do contexto")

# ITERADORES
class Contador:
    def __init__(self, max):
        self.max = max
        self.n = 0
    
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.n < self.max:
            resultado = self.n
            self.n += 1
            return resultado
        raise StopIteration

print("\nIterador personalizado:")
for num in Contador(5):
    print(num, end=" ")

# MÓDULOS ÚTEIS
import datetime
import random
from collections import Counter, defaultdict, deque
from itertools import combinations, permutations

print(f"\n\nData atual: {datetime.datetime.now()}")
print(f"Número aleatório: {random.randint(1, 100)}")

palavras = ['python', 'java', 'python', 'c++']
contagem = Counter(palavras)
print(f"Contagem: {contagem}")

combos = list(combinations([1, 2, 3], 2))
print(f"Combinações: {combos}")

# EXPRESSÕES REGULARES
import re

texto = "Meu email é joao@example.com"
padrao = r'[\w.-]+@[\w.-]+\.\w+'
email = re.search(padrao, texto)

if email:
    print(f"\nEmail encontrado: {email.group()}")

# Substituição
texto2 = "Telefone: 123-456-7890"
novo_texto = re.sub(r'\d', 'X', texto2)
print(f"Mascarado: {novo_texto}")

# PROGRAMAÇÃO FUNCIONAL
from functools import reduce

nums = [1, 2, 3, 4, 5]

# map
dobros = list(map(lambda x: x*2, nums))
print(f"\nDobros: {dobros}")

# filter
pares = list(filter(lambda x: x % 2 == 0, nums))
print(f"Pares: {pares}")

# reduce
produto = reduce(lambda x, y: x * y, nums)
print(f"Produto: {produto}")

# zip
nomes = ['Ana', 'Bob', 'Carol']
idades = [25, 30, 35]
combinados = list(zip(nomes, idades))
print(f"Zip: {combinados}")

# enumerate
for i, (nome, idade) in enumerate(zip(nomes, idades), 1):
    print(f"{i}. {nome}: {idade} anos")
```

```python
# EXEMPLO 1: Sistema de Biblioteca
class Livro:
    def __init__(self, titulo, autor, ano):
        self.titulo = titulo
        self.autor = autor
        self.ano = ano
        self.emprestado = False
    
    def emprestar(self):
        if not self.emprestado:
            self.emprestado = True
            return f"{self.titulo} foi emprestado"
        return f"{self.titulo} já está emprestado"
    
    def devolver(self):
        if self.emprestado:
            self.emprestado = False
            return f"{self.titulo} foi devolvido"
        return f"{self.titulo} não estava emprestado"
    
    def __str__(self):
        status = "Emprestado" if self.emprestado else "Disponível"
        return f"{self.titulo} - {self.autor} ({self.ano}) [{status}]"

class Biblioteca:
    def __init__(self):
        self.livros = []
    
    def adicionar_livro(self, livro):
        self.livros.append(livro)
    
    def buscar_por_titulo(self, titulo):
        for livro in self.livros:
            if titulo.lower() in livro.titulo.lower():
                return livro
        return None
    
    def listar_disponiveis(self):
        return [l for l in self.livros if not l.emprestado]
    
    def __str__(self):
        return "\n".join(str(l) for l in self.livros)

# Usando o sistema
bib = Biblioteca()
bib.adicionar_livro(Livro("1984", "George Orwell", 1949))
bib.adicionar_livro(Livro("Python Fluente", "Luciano Ramalho", 2015))
bib.adicionar_livro(Livro("Clean Code", "Robert Martin", 2008))

print("=== BIBLIOTECA ===")
print(bib)

livro = bib.buscar_por_titulo("1984")
if livro:
    print(f"\n{livro.emprestar()}")

print("\n=== LIVROS DISPONÍVEIS ===")
for l in bib.listar_disponiveis():
    print(l)
```

```python
# EXEMPLO 2: Análise de Dados Simples
class AnalisadorDados:
    def __init__(self, dados):
        self.dados = dados
    
    def media(self):
        return sum(self.dados) / len(self.dados) if self.dados else 0
    
    def mediana(self):
        sorted_data = sorted(self.dados)
        n = len(sorted_data)
        meio = n // 2
        if n % 2 == 0:
            return (sorted_data[meio-1] + sorted_data[meio]) / 2
        return sorted_data[meio]
    
    def moda(self):
        from collections import Counter
        contagem = Counter(self.dados)
        max_freq = max(contagem.values())
        return [k for k, v in contagem.items() if v == max_freq]
    
    def desvio_padrao(self):
        media = self.media()
        variancia = sum((x - media) ** 2 for x in self.dados) / len(self.dados)
        return variancia ** 0.5
    
    def relatorio(self):
        return f"""
=== RELATÓRIO ESTATÍSTICO ===
Dados: {self.dados}
Quantidade: {len(self.dados)}
Mínimo: {min(self.dados)}
Máximo: {max(self.dados)}
Média: {self.media():.2f}
Mediana: {self.mediana():.2f}
Moda: {self.moda()}
Desvio Padrão: {self.desvio_padrao():.2f}
"""

# Teste
dados = [10, 15, 10, 20, 25, 30, 10, 15, 20, 25]
analisador = AnalisadorDados(dados)
print(analisador.relatorio())
```

```python
# EXEMPLO 3: Validador de CPF
def validar_cpf(cpf):
    """Valida CPF brasileiro"""
    # Remove caracteres não numéricos
    cpf = ''.join(filter(str.isdigit, cpf))
    
    # Verifica se tem 11 dígitos
    if len(cpf) != 11:
        return False
    
    # Verifica se todos os dígitos são iguais
    if cpf == cpf[0] * 11:
        return False
    
    # Calcula o primeiro dígito verificador
    soma = sum(int(cpf[i]) * (10 - i) for i in range(9))
    digito1 = (soma * 10 % 11) % 10
    
    # Calcula o segundo dígito verificador
    soma = sum(int(cpf[i]) * (11 - i) for i in range(10))
    digito2 = (soma * 10 % 11) % 10
    
    # Verifica se os dígitos calculados correspondem aos informados
    return cpf[-2:] == f"{digito1}{digito2}"

# Testes
cpfs_teste = [
    "123.456.789-09",
    "111.111.111-11",
    "529.982.247-25"
]

print("=== VALIDAÇÃO DE CPF ===")
for cpf in cpfs_teste:
    valido = "✓ Válido" if validar_cpf(cpf) else "✗ Inválido"
    print(f"{cpf}: {valido}")
```

```python
# EXERCÍCIO 1: Criar função que retorna os N primeiros números da sequência Fibonacci
def fibonacci(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    
    fib = [0, 1]
    for i in range(2, n):
        fib.append(fib[i-1] + fib[i-2])
    return fib

print("Fibonacci(10):", fibonacci(10))

# EXERCÍCIO 2: Contar palavras em um texto
def contar_palavras(texto):
    palavras = texto.lower().split()
    from collections import Counter
    return Counter(palavras)

texto = "Python é incrível Python é poderoso Python"
print("\nContagem:", contar_palavras(texto))

# EXERCÍCIO 3: Verificar se uma string é palíndromo
def eh_palindromo(texto):
    texto = ''.join(texto.lower().split())
    return texto == texto[::-1]

print("\n'arara' é palíndromo?", eh_palindromo("arara"))
print("'python' é palíndromo?", eh_palindromo("python"))

# EXERCÍCIO 4: Filtrar números primos de uma lista
def eh_primo(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def filtrar_primos(lista):
    return [n for n in lista if eh_primo(n)]

numeros = list(range(1, 30))
print(f"\nPrimos até 30: {filtrar_primos(numeros)}")

# EXERCÍCIO 5: Ordenar lista de dicionários por múltiplos critérios
alunos = [
    {'nome': 'Ana', 'nota': 8.5, 'idade': 20},
    {'nome': 'Bruno', 'nota': 9.0, 'idade': 19},
    {'nome': 'Carlos', 'nota': 8.5, 'idade': 21},
    {'nome': 'Diana', 'nota': 9.0, 'idade': 20}
]

# Ordenar por nota (desc) e depois por idade (asc)
ordenados = sorted(alunos, key=lambda x: (-x['nota'], x['idade']))

print("\nAlunos ordenados:")
for aluno in ordenados:
    print(f"{aluno['nome']}: {aluno['nota']} - {aluno['idade']} anos")
```
