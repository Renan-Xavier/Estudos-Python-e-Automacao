# Exemplo de Listas em Python

As listas são uma das estruturas de dados mais versáteis e comumente usadas em Python. Vou mostrar um exemplo prático que demonstra várias operações com listas.


```python
# Criando uma lista
frutas = ["maçã", "banana", "laranja", "uva", "manga"]

# Exibindo a lista completa
print("Lista de frutas:", frutas)

# Acessando elementos por índice
print("Primeira fruta:", frutas[0])
print("Última fruta:", frutas[-1])

# Fatiando a lista
print("Duas primeiras frutas:", frutas[0:2])

# Adicionando elementos
frutas.append("abacaxi")
print("Lista após adicionar abacaxi:", frutas)

# Inserindo em posição específica
frutas.insert(2, "morango")
print("Lista após inserir morango na posição 2:", frutas)

# Removendo elementos
frutas.remove("banana")
print("Lista após remover banana:", frutas)

# Removendo por índice
fruta_removida = frutas.pop(1)
print(f"Fruta removida: {fruta_removida}")
print("Lista após remover pelo índice 1:", frutas)

# Verificando se um elemento existe na lista
if "uva" in frutas:
    print("Uva está na lista!")

# Comprimento da lista
print("Número de frutas na lista:", len(frutas))

# Ordenando a lista
frutas.sort()
print("Lista ordenada:", frutas)

# Invertendo a lista
frutas.reverse()
print("Lista invertida:", frutas)

# Criando uma cópia da lista
frutas_copia = frutas.copy()
print("Cópia da lista:", frutas_copia)

# Limpando a lista
frutas.clear()
print("Lista após limpar:", frutas)
print("A cópia permanece intacta:", frutas_copia)
```


Este exemplo demonstra as operações mais comuns com listas em Python, incluindo:
- Criação de listas
- Acesso a elementos
- Adição e remoção de elementos
- Verificação de existência
- Ordenação e inversão
- Cópia e limpeza

Outros pacotes Python úteis para trabalhar com listas de forma mais avançada incluem:
- `numpy` para operações numéricas eficientes com arrays
- `pandas` para manipulação de dados tabulares
- `collections` (da biblioteca padrão) que oferece estruturas de dados especializadas como `deque` e `Counter`

# Exemplo com NumPy e Pandas

Vou criar um exemplo que demonstra como trabalhar com listas usando NumPy e Pandas, duas bibliotecas poderosas para manipulação de dados em Python.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ---- PARTE 1: NUMPY ----
print("===== EXEMPLOS COM NUMPY =====")

# Criando arrays NumPy a partir de listas
lista_numeros = [10, 25, 30, 45, 50, 65, 70, 85, 90]
array_np = np.array(lista_numeros)
print("Array NumPy:", array_np)

# Operações matemáticas vetorizadas (muito mais rápidas que com listas)
print("Média:", np.mean(array_np))
print("Soma:", np.sum(array_np))
print("Desvio padrão:", np.std(array_np))

# Reshape - transformando em matriz 3x3
matriz = array_np.reshape(3, 3)
print("Matriz 3x3:\n", matriz)

# Operações com matrizes
print("Transposta:\n", matriz.T)
print("Diagonal:", np.diag(matriz))

# Criando arrays especiais
print("Array de zeros:", np.zeros(5))
print("Array de uns:", np.ones(5))
print("Array com valores aleatórios:", np.random.random(5))
print("Range de valores:", np.arange(0, 10, 2))  # de 0 a 10, passo 2

# Operações matemáticas em massa
print("Raiz quadrada de cada elemento:", np.sqrt(array_np))
print("Exponencial de cada elemento:", np.exp(array_np / 10))

# ---- PARTE 2: PANDAS ----
print("\n===== EXEMPLOS COM PANDAS =====")

# Criando um DataFrame a partir de um dicionário de listas
dados = {
    'Nome': ['Ana', 'Bruno', 'Carlos', 'Daniela', 'Eduardo'],
    'Idade': [25, 32, 18, 45, 37],
    'Cidade': ['São Paulo', 'Rio de Janeiro', 'Belo Horizonte', 'Brasília', 'Salvador'],
    'Salário': [3500, 4200, 2800, 5100, 3900]
}

df = pd.DataFrame(dados)
print("DataFrame:\n", df)

# Informações básicas
print("\nInformações do DataFrame:")
print(df.info())
print("\nEstatísticas descritivas:")
print(df.describe())

# Selecionando colunas
print("\nApenas nomes e idades:")
print(df[['Nome', 'Idade']])

# Filtrando dados
print("\nPessoas com mais de 30 anos:")
print(df[df['Idade'] > 30])

# Ordenando dados
print("\nOrdenado por salário (decrescente):")
print(df.sort_values('Salário', ascending=False))

# Agrupamento e agregação
print("\nMédia de salário por cidade:")
print(df.groupby('Cidade')['Salário'].mean())

# Adicionando uma nova coluna
df['Bônus'] = df['Salário'] * 0.1
print("\nDataFrame com coluna de bônus:")
print(df)

# Aplicando função a uma coluna
df['Faixa Etária'] = df['Idade'].apply(lambda x: 'Jovem' if x < 30 else 'Adulto' if x < 40 else 'Sênior')
print("\nDataFrame com faixa etária:")
print(df)

# Salvando e carregando dados
# df.to_csv('dados_funcionarios.csv', index=False)
# df_carregado = pd.read_csv('dados_funcionarios.csv')

# Visualização básica (descomente para ver o gráfico)
plt.figure(figsize=(10, 6))
df.plot(kind='bar', x='Nome', y='Salário')
plt.title('Salários por Funcionário')
plt.tight_layout()
plt.show()
```


Este exemplo demonstra:

**Com NumPy:**
- Criação de arrays a partir de listas
- Operações matemáticas vetorizadas (muito mais eficientes que listas padrão)
- Manipulação de matrizes
- Funções matemáticas aplicadas a todos os elementos de uma vez

**Com Pandas:**
- Criação de DataFrames a partir de dicionários
- Análise exploratória básica de dados
- Filtragem e seleção de dados
- Ordenação e agrupamento
- Adição de novas colunas e aplicação de funções
- Exportação e importação de dados

Estas bibliotecas são essenciais para análise de dados e ciência de dados em Python, oferecendo funcionalidades muito mais poderosas e eficientes do que as listas nativas do Python para conjuntos de dados maiores.
