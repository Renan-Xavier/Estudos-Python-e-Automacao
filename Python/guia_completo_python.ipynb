## Operadores

Operadores permitem combinar valores e produzir resultados. Os principais grupos:

- **Aritméticos**: `+`, `-`, `*`, `/`, `//`, `%`, `**`
- **Comparação**: `==`, `!=`, `>`, `<`, `>=`, `<=`
- **Lógicos**: `and`, `or`, `not`
- **Atribuição composta**: `+=`, `-=`, `*=`, etc.

Python respeita precedência matemática. Use parênteses para deixar a intenção clara.

````python
a, b = 10, 3
print("Soma:", a + b)
print("Divisão real:", a / b)
print("Divisão inteira:", a // b)
print("Resto:", a % b)
print("Potência:", a ** b)

print("Maior que:", a > b)
print("Igualdade:", a == 10)
print("Combinação lógica:", (a > b) and (b > 0))
````

## Coleções (estruturas de dados)

Python oferece coleções flexíveis e poderosas:

- **Listas (`list`)**: mutáveis, ordenadas
- **Tuplas (`tuple`)**: imutáveis, ordenadas
- **Conjuntos (`set`)**: não ordenados, elementos únicos
- **Dicionários (`dict`)**: pares chave-valor

Conhecer quando usar cada coleção é essencial para escrever código eficiente.

````python
lista = [1, 2, 3]
lista.append(4)
print("Lista:", lista)

ponto = (10, 20)
print("Tupla:", ponto)

numeros = {1, 2, 2, 3}
print("Set (valores únicos):", numeros)

aluno = {"nome": "Ana", "idade": 20}
aluno["curso"] = "Python"
print("Dicionário:", aluno)
````

### Operações frequentes com coleções

- Listas e tuplas suportam slicing (`colecao[inicio:fim:passo]`).
- Sets permitem operações matemáticas (`union`, `intersection`).

## Condicionais e loops

Fluxo de controle permite executar blocos de código de forma seletiva ou repetitiva.

- `if`, `elif`, `else`
- `for` para iterar sobre coleções
- `while` para repetir até que uma condição seja falsa
- Palavras-chave úteis: `break`, `continue`, `enumerate`, `range`


````python
nota = 7.5

if nota >= 7:
    situacao = "Aprovado"
elif nota >= 5:
    situacao = "Recuperação"
else:
    situacao = "Reprovado"

print("Situação:", situacao)

for indice, valor in enumerate([10, 20, 30], start=1):
    print(indice, valor)

contador = 0
while contador < 3:
    print("while contador:", contador)
    contador += 1
````

## Funções

Funções permitem reutilizar lógica e organizar o código.

- Definidas com `def`
- Parâmetros posicionais, nomeados e padrão
- Argumentos variáveis (`*args`, `**kwargs`)
- Retorno com `return`

Documente funções com docstrings curtas e objetivas.

````python
def calcular_media(*notas, arredondar=False):
    """Calcula a média aritmética das notas fornecidas."""
    if not notas:
        raise ValueError("Informe pelo menos uma nota")
    media = sum(notas) / len(notas)
    return round(media, 2) if arredondar else media

print(calcular_media(7, 8, 9))
print(calcular_media(7, 8, 9, arredondar=True))
````

### Funções anônimas (lambda) e ordem superior

Funções lambda são expressões curtas; ordem superior significa que funções podem receber ou retornar outras funções.

````python
valores = [1, 2, 3, 4]
quadrados = list(map(lambda x: x ** 2, valores))
pares = list(filter(lambda x: x % 2 == 0, valores))
print("Quadrados:", quadrados)
print("Pares:", pares)
````

## Escopo, closures e boas práticas com funções

- Variáveis definidas dentro de funções ficam em escopo local.
- `global` e `nonlocal` alteram comportamentos, mas use-os com cautela.
- Closures capturam valores do escopo externo.


````python
def contador():
    total = 0

    def incrementar():
        nonlocal total
        total += 1
        return total

    return incrementar

c = contador()
print(c())
print(c())
````

## Entrada e saída de arquivos

Utilize `with open(caminho, modo, encoding)` para ler ou escrever arquivos. O `with` garante o fechamento automático.

Modos comuns: `"r"` (leitura), `"w"` (escrita, sobrescreve), `"a"` (anexa), `"r+"` (leitura e escrita).

````python
from pathlib import Path

caminho = Path("exemplo.txt")

with caminho.open("w", encoding="utf-8") as arquivo:
    arquivo.write("Linha 1
")
    arquivo.write("Linha 2
")

with caminho.open("r", encoding="utf-8") as arquivo:
    conteudo = arquivo.read()

print(conteudo)
````

## Tratamento de exceções

Use `try`/`except` para capturar erros, `else` para código que roda apenas se não houve exceção e `finally` para código que deve sempre executar.

Crie exceções personalizadas herdando de `Exception` quando precisar representar erros de domínio.

````python
def dividir(a, b):
    try:
        resultado = a / b
    except ZeroDivisionError as erro:
        print("Não é possível dividir por zero:", erro)
    else:
        print("Resultado:", resultado)
    finally:
        print("Operação finalizada")

dividir(10, 2)
dividir(10, 0)
````

## Programação funcional e iteráveis

A biblioteca padrão oferece ferramentas para trabalhar com iteradores:

- `itertools`: combinações, permutações, contagem infinita
- `functools`: utilitários como `reduce`, `partial`, `lru_cache`
- Geradores (`yield`) criam iteradores de forma eficiente.

````python
from functools import reduce
from itertools import permutations

numeros = [1, 2, 3, 4]
produto = reduce(lambda x, y: x * y, numeros)
print("Produto:", produto)

print("Permutações de 'abc':")
for p in permutations("abc", 2):
    print(p)
````

## Comprehensions

Comprehensions permitem construir coleções de forma concisa:

- List comprehension: `[expressao for item in iteravel]`
- Dict comprehension: `{chave: valor for item in iteravel}`
- Set comprehension: `{expressao for item in iteravel}`
- Generator expression: `(expressao for item in iteravel)`


````python
numeros = [1, 2, 3, 4, 5]
quadrados = [n ** 2 for n in numeros]
pares = {n for n in numeros if n % 2 == 0}
indice_por_numero = {n: i for i, n in enumerate(numeros)}

print(quadrados)
print(pares)
print(indice_por_numero)
````

## Programação orientada a objetos (POO)

POO permite modelar entidades usando classes. Conceitos principais:

- Classe: define atributos e métodos
- Objeto: instância da classe
- Encapsulamento: esconder detalhes internos
- Herança: reutilizar comportamentos
- Polimorfismo: comportamentos diferentes usando a mesma interface


````python
class ContaBancaria:
    taxa_juros = 0.01

    def __init__(self, titular, saldo=0):
        self.titular = titular
        self._saldo = saldo

    def depositar(self, valor):
        self._saldo += valor

    def sacar(self, valor):
        if valor > self._saldo:
            raise ValueError("Saldo insuficiente")
        self._saldo -= valor

    def aplicar_juros(self):
        self._saldo += self._saldo * self.taxa_juros

    def __repr__(self):
        return f"ContaBancaria(titular={self.titular!r}, saldo={self._saldo:.2f})"


class ContaPremium(ContaBancaria):
    taxa_juros = 0.02

    def aplicar_cashback(self, percentual):
        self._saldo += self._saldo * percentual


conta = ContaPremium("Ana", saldo=1000)
conta.aplicar_juros()
conta.aplicar_cashback(0.05)
print(conta)
````

### Classes abstratas e protocolos

Use `abc.ABC` para definir interfaces. Em Python 3.8+, `typing.Protocol` descreve comportamentos esperados sem herança explícita.

````python
from abc import ABC, abstractmethod
from typing import Protocol


class Persistivel(ABC):
    @abstractmethod
    def salvar(self) -> None:
        raise NotImplementedError


class Exportavel(Protocol):
    def exportar(self) -> str:
        ...


class Relatorio(Persistivel):
    def __init__(self, dados):
        self.dados = dados

    def salvar(self) -> None:
        print("Relatório salvo:", self.dados)

    def exportar(self) -> str:
        return ";".join(self.dados)


relatorio = Relatorio(["coluna1", "coluna2"])
relatorio.salvar()
print(relatorio.exportar())
````

## Recursos avançados da linguagem

Esta seção reúne funcionalidades poderosas para escrever código mais expressivo.

- **Decoradores**: funções que recebem outra função e retornam uma versão modificada
- **Geradores**: funções com `yield`, criam iteradores sob demanda
- **Context managers**: controlam entrada/saída de recursos com `with`
- **Dataclasses**: reduzem boilerplate em classes de dados
- **Type hints**: anotações de tipo opcionais, auxiliam ferramentas e legibilidade


````python
from functools import wraps


def medir_tempo(funcao):
    @wraps(funcao)
    def wrapper(*args, **kwargs):
        import time
        inicio = time.perf_counter()
        resultado = funcao(*args, **kwargs)
        duracao = time.perf_counter() - inicio
        print(f"{funcao.__name__} executou em {duracao:.6f}s")
        return resultado

    return wrapper


@medir_tempo
def somar_numeros(n: int) -> int:
    return sum(range(n))

print(somar_numeros(1_000_000))
````

````python
def gerar_fibonacci(limite):
    anterior, atual = 0, 1
    while atual <= limite:
        yield atual
        anterior, atual = atual, anterior + atual

print(list(gerar_fibonacci(50)))
````

````python
from contextlib import contextmanager


@contextmanager
def abrir_arquivo(caminho, modo):
    arquivo = open(caminho, modo, encoding="utf-8")
    try:
        yield arquivo
    finally:
        arquivo.close()


with abrir_arquivo("contexto.txt", "w") as arquivo:
    arquivo.write("Usando um context manager personalizado
")
````

### Anotações de tipo (type hints)

Type hints não mudam o comportamento em tempo de execução, mas ajudam a detectar erros com ferramentas (`mypy`, `pyright`). Use-os para documentar intenções.


````python
from typing import Iterable, Union


def soma_total(valores: Iterable[Union[int, float]]) -> float:
    total: float = 0
    for valor in valores:
        total += float(valor)
    return total

print(soma_total([1, 2.5, 3]))
````

## Concorrência e paralelismo básico

Python oferece diferentes abordagens para lidar com tarefas concorrentes:

- `threading`: múltiplas threads, bom para IO-bound
- `multiprocessing`: processos separados, útil para CPU-bound
- `asyncio`: programação assíncrona baseada em *event loop*

A escolha depende do tipo de carga e das limitações do GIL (Global Interpreter Lock).

````python
import asyncio
import time


async def tarefa(nome, duracao):
    print(f"Iniciando {nome}")
    await asyncio.sleep(duracao)
    print(f"Finalizando {nome}")


async def main():
    inicio = time.perf_counter()
    await asyncio.gather(
        tarefa("A", 1),
        tarefa("B", 2),
        tarefa("C", 1.5),
    )
    duracao = time.perf_counter() - inicio
    print(f"Tarefas concluídas em {duracao:.2f}s")


asyncio.run(main())
````

## Boas práticas, testes e logging

- Siga a PEP 8 (estilo oficial). Ferramentas: `black`, `flake8`, `pylint`.
- Escreva testes automatizados (unittest, pytest) para prevenir regressões.
- Utilize `logging` para registrar eventos, evitando imprimir diretamente em aplicações complexas.


````python
import logging
import unittest


logging.basicConfig(level=logging.INFO)


def somar(a, b):
    logging.info("Somando %s e %s", a, b)
    return a + b


class TestSomar(unittest.TestCase):
    def test_soma_inteiros(self):
        self.assertEqual(somar(2, 3), 5)

    def test_soma_float(self):
        self.assertAlmostEqual(somar(2.5, 3.5), 6.0)


suite = unittest.TestLoader().loadTestsFromTestCase(TestSomar)
resultado = unittest.TextTestRunner(verbosity=2).run(suite)
print("Falhas:", len(resultado.failures))
````
